````python
print("This is the printing using the double quotes")
print('This is the printing using the single quotes')
````

````python
# We have five charachters that should be printed using the escape character
print('This is the print of double quotes \"')
print("This is the print of single quotes \'")
print("This is the print of back slash \\")
print("This is the print of the new \nline")
print("This is the print of tab \tcharacter")
````
